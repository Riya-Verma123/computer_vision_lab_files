```python
import cv2
import matplotlib.pyplot as plt

# Path to your image
image_path = "/content/sea.jpeg"

# Read the image
image = cv2.imread(image_path, cv2.IMREAD_COLOR)

# Check if image was loaded
if image is None:
    print("Error: Could not load image.")
else:
    print("Image loaded successfully!")

    # Convert BGR image to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Convert color image to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Convert grayscale image to binary
    _, binary_image = cv2.threshold(
        gray_image,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Convert grayscale image to complement
    complement_image = cv2.bitwise_not(gray_image)

    # Display all four images
    plt.figure(figsize=(20, 5))

    # Original Color Image
    plt.subplot(1, 4, 1)
    plt.imshow(image_rgb)
    plt.title("Original Color Image")
    plt.axis("off")

    # Grayscale Image
    plt.subplot(1, 4, 2)
    plt.imshow(gray_image, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Binary Image
    plt.subplot(1, 4, 3)
    plt.imshow(binary_image, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    # Complement Image
    plt.subplot(1, 4, 4)
    plt.imshow(complement_image, cmap="gray")
    plt.title("Complement Image")
    plt.axis("off")

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

image_path = "/content/sea.jpeg"

image = cv2.imread(image_path, cv2.IMREAD_COLOR)

if image is None:
    print("Error: Could not load image.")
else:
    print("Image loaded successfully!")

    # Convert original image from BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # ---------------- RESIZE ----------------
    resized_image = cv2.resize(image, (400, 300))
    resized_rgb = cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)

    # ---------------- CROP ----------------
    cropped_image = image[50:400, 50:500]
    cropped_rgb = cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)

    # ---------------- ROTATE ----------------
    rotated_image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    rotated_rgb = cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB)

    # ---------------- DISPLAY ----------------
    plt.figure(figsize=(15, 10))

    # Original Image
    plt.subplot(2, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Resized Image
    plt.subplot(2, 2, 2)
    plt.imshow(resized_rgb)
    plt.title("Resized Image")
    plt.axis("off")

    # Cropped Image
    plt.subplot(2, 2, 3)
    plt.imshow(cropped_rgb)
    plt.title("Cropped Image")
    plt.axis("off")

    # Rotated Image
    plt.subplot(2, 2, 4)
    plt.imshow(rotated_rgb)
    plt.title("Rotated Image")
    plt.axis("off")

    plt.show()
```
